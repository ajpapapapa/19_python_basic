```python
#for 와 range 사용하기
a = [
    [10,20],
    [30,40],
    [50,60]
]

for i in range(len(a)):         # len(a)는 3이므로 range(3) -> i는 0, 1, 2
    for j in range(len(a[i])):  # 각 줄의 길이는 2이므로 range(2) -> j는 0, 1
        print(a[i][j], end=" ")
    print()
```

```python
# 23.7 심사문제: 지뢰찾기
# 표준 입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그 다음 줄부터
# 리스트의 요소로 들어갈 문자가 입력됩니다. 이때 2차원 리스트 안에서 *는 지뢰이고
#  .은 지뢰가 아닙니다. 지뢰가 아닌 요소에는 인접한 지뢰의 개수를 출력하는 프로그램을 만드세요
# 여러 줄을 입력 받으려면 다음과 같이 for 반복문에서 input을 호출한 뒤 append로 각 줄을
# 추가하면 됩니다(list 안에 문자열을 넣으면 문자열이 문자 리스트로 변환됩니다).

col, row = map(int, input().split())
# 가로크기 column, 세로크기 row

matrix = [] #지뢰(*)와 빈칸(.)들이 담길 빈 2차원 리스트(모눈종이 판)를 준비합니다.
for i in range(row): 
    matrix.append(list(input()))

# 결과를 출력하기 위해 입력받은 matrix를 그대로 복사해서 빈 판을 만듭니다.
# (지뢰가 아닐 때 숫자로 바꿔주기 위함)
result = [row[:] for row in matrix]

for r in range(row):
    for c in range(col):
        # 현재 칸이 지뢰(*)면 탐색할 필요가 없으므로 건너뜁니다.
        # 바깥쪽 루프(r)가 한 줄씩 내려가면서, 안쪽 루프(c)가
        # 그 줄의 왼쪽부터 오른쪽 끝까지 모든 칸을 1칸씩 빠짐없이 검사합니다.
        if matrix[r][c] == '*':
            continue
            # 지금 검사하고 있는 칸(matrix[r][c])에 지뢰(*)가 박혀있다면?
            # 지뢰 자리에 숫자를 적을 필요가 없으므로, 뒤의 복잡한 계산을 할 필요 없이 
            # 곧바로 다음 칸으로 넘어갑니다(continue).
        count = 0 # 발견한 지뢰 개수를 셀 변수 초기화
        
        # 8방향 탐색 (세로 -1~1, 가로 -1~1)
        for dr in [-1, 0, 1]: #dr은 세로 방향으로 이동(-1은 위, 0은 가만히, 1은 아래)
            for dc in [-1, 0, 1]: #dc는 가로 방향으로 이동(-1은 왼쪽, 0은 가만히, 1은 오른쪽)
                # 자기 자신(0, 0)은 지뢰 개수에 포함 안 함
                if dr == 0 and dc == 0:
                    continue
                
                nr = r + dr
                nc = c + dc
                
                # 리스트 범위를 벗어나지 않는지 안전장치 확인
                if 0 <= nr < row and 0 <= nc < col: #만약 맨 모서리(구석) 칸에 있을 때 위쪽(-1)을 보려고 하면 모눈종이 밖으로 튀어나가 에러가 납니다. 그걸 막기 위해 "판 내부 범위에 안전하게 들어올 때만 검사해라"라는 방어막을 쳐주는 것입니다.
                    if matrix[nr][nc] == '*': #안전한 범위 안에서 주변 칸을 열어봤더니 지뢰(*)가 들어있다면, 지뢰 개수(count)를 1개씩 꾹꾹 눌러 담습니다.
                        count += 1
                        
        # 세어낸 지뢰 개수를 문자열로 변환하여 결과 판에 저장
        result[r][c] = str(count)

# 최종 결과를 문제에서 원하는 형태로 출력 (각 행을 이어붙여서 출력)
for r in range(row):
    print(''.join(result[r]))
```

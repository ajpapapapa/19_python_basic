```python
# 2차원 리스트 만들고 요소에 접근하기
a = [
    [10,20], [30,40], [50,60]
]
a
a[0][0] # 세로인덱스 0, 가로인덱스 0 인 요소 10 출력
a[1][1] # 세로인덱스 1, 가로인덱스 1 인 요소 40 출력

a[0][1] = 1000 # 세로인덱스 0, 가로인덱스 1 인 요소 20을 1000으로 변경
print(a)
```

```python
#반목문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10,20],
    [30,40],
    [50,60]
]

for x,y in a:
    print(x,y)
```

```python
#반목문 두번 사용하기
a = [
    [10,20],
    [30,40],
    [50,60]
]

for i in a:
    for j in i:
        print(j, end=" ")
    print()
```

```python
#for 와 range 사용하기
a = [
    [10,20],
    [30,40],
    [50,60]
]

for i in range(len(a)):         # len(a)는 3이므로 range(3) -> i는 0, 1, 2
    for j in range(len(a[i])):  # 각 줄의 길이는 2이므로 range(2) -> j는 0, 1
        print(a[i][j], end=" ")
    print()
```

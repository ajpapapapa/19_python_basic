```python
# while 반복문 사용하기
# 초기식
# while 조건식:
#     반복할코드
#     변화식

i = 0
while i < 100:
    print('Hello, world!')
    i += 1
```

```python
# 초깃값을 1부터 시작하기
i = 1
while i <= 10:
    print('Hello, world!',i)
    i += 1
```

```python
# 초깃값을 감소시키기
i = 10
while i > 0:
    print('Hello, world!',i)
    i -= 1
```

```python
# 입력한 횟수대로 반복하기

count = int(input())
i = 0
while i < count:
    print("Hello, world!", i)
    i += 1
```

```python
import random #랜덤 모듈을 가지고 옴
i=0
while i != 3: # 3이 아닐 때 계속 반복
    i = random.randint(1,6) # randint을 사용하여 1과 6 사이의 난수를 생성
    print(i)
```

```python
# 17.5 연습문제: 변수 두 개를 다르게 반복하기

#다음 소스 코드를 완성하여 정수 2 5, 4 4, 8 3, 16 2, 32 1이
#각 줄에 출력되게 만드세요. while에 조건식은 두 개 지정하고,
#두 변수를 모두 변화시켜야 합니다.

i = 2
j = 5

while i <= 32 or j >= 1:
    print(i,j)
    i *= 2
    j -= 1
```

```python
# 17.6 심사문제: 교통카드 잔액 출력하기

#표준 입력으로 금액(정수)이 입력됩니다. 1회당 요금은 1,350원이고,
#교통카드를 사용했을 때마다의 잔액을 각 줄에 출력하는 프로그램을 만드세요
#단, 최초 금액은 출력하지 않아야 합니다.
# 그리고 잔액은 음수가 될 수 없으며 잔액이 부족하면 출력을 끝냅니다.

balance = int(input()) # 최초 금액을 정수형으로 입력받기


while balance >= 1350: # 잔액이 1회 요금(1350원) 이상일 동안 반복
    balance -= 1350       # 요금 차감
    print(balance)        # 차감된 잔액 출력
```

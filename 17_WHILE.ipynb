```python
# while 반복문 사용하기
# 초기식
# while 조건식:
#     반복할코드
#     변화식

i = 0
while i < 100:
    print('Hello, world!')
    i += 1
```

```python
# 초깃값을 1부터 시작하기
i = 1
while i <= 10:
    print('Hello, world!',i)
    i += 1
```

```python
# 초깃값을 감소시키기
i = 10
while i > 0:
    print('Hello, world!',i)
    i -= 1
```

```python
# 입력한 횟수대로 반복하기

count = int(input())
i = 0
while i < count:
    print("Hello, world!", i)
    i += 1
```

```python
# 반복 횟수가 정해지지 않은 경우

import random

print(random.random())
print(random.random())
random.randint(1,6)
```

```python
import random #랜덤 모듈을 가지고 옴
i=0
while i != 3: # 3이 아닐 때 계속 반복
    i = random.randint(1,6) # randint을 사용하여 1과 6 사이의 난수를 생성
    print(i)
```

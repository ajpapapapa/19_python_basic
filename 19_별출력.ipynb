```python
# 중첩루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep="",end=" ")
    print("i:", i, "\\n", sep="")
```

```python
# 사격형으로 별 출력하기
for i in range(5): # 1번 바깥쪽 반복문 (세로 방향 담당)
    for j in range(5): # 2번 안쪽 반복문 (가로 방향 담당)
        print('*',end='') # 별을 옆으로 나란히 출력
    print() # 줄바꿈 담당
```

```python
# 계단식으로 별 출력하기
for i in range(5):
    for j in range(5): # i=0일때 j=0, i=1 일때 j=0,j=1 이니까 이런식으로 계단식
        if j <= i:
            print('*',end='') 
    print()
```

```python
# 대각선으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j == i:
            print('*',end='') 
        else:
            print(' ', end='')
    print()
```

```python
# 19.5 연습문제: 역삼각형 모양으로 별 출력하기
# 다음 소스 코드를 완성하여 역삼각형 모양으로 별이 출력되게 만드세요.

for i in range(5):
    for j in range(5):
        if j < i:
            print(' ', end='')
        else:
            print('*', end='')                         
    print()
```

```python
# 19.6 심사문제: 산 모양으로 별 출력하기
# 표준 입력으로 삼각형의 높이가 입력됩니다. 입력된 높이만큼 산 모양으로 별을
# 출력하는 프로그램을 만드세요. 이때 출력결과는 예제와 정확히 일치해야 합니다.
# 모양이 같더라도 공백이나 빈 줄이 더 들어가면 틀린 것으로 처리됩니다.

h = int(input())
 
for i in range(h):
    for j in range(h-1-i): # 1. 공백 출력 (왼쪽 여백)
        print(' ', end='')
    for j in range(2*i+1): # 2. 별 출력 (홀수 개씩 증가)
        print('*', end='')                         
    print()
```

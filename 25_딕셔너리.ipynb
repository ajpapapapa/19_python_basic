```python
#딕셔너리에 키와 기본값 저장하기

x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}

x
```

```python
x.setdefault("e")
x
```

```python
x.setdefault("f", 100)
x
```

```python
#딕셔너리에 키의 값 수정하기

x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}

x.update(a=90)
x
```

```python
#딕셔너리에 키-값 쌍 삭제하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.pop('a')
x
```

```python
x.clear()
x
```

```python
#딕셔너리에 키의 값을 가져오기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
x.get('a')
```

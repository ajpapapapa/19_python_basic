```python
# 심사문제 : 온라인 할인 쿠폰 시스템 만들기
# 표준 입력으로 가격(정수)과 쿠폰 이름이 각 줄에 입력됩니다. cash3000 쿠폰은 3000원,
# cash5000 쿠폰은 5000원 할인합니다.쿠폰에 따라 할인된 가격을 출력하는 프로그램을 만드세요.

x = int(input())
y = input()

if y == 'cash3000':
    x -= 3000 # x = x-3000 과 같은 말

if y == 'cash5000':
    x -= 5000

print(x)
```

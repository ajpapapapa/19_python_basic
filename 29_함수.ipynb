```python
# 함수에서 값을 여러개 반환하기
def add_sub(a,b):
    return a+b, a-b

result = add_sub(10,20)
print(result)
```

```python
# 연습문제: 몫과 나머지를 구하는 함수 만들기

x = 10
y = 3

def get_quotient_remainer(a,b):
    return a//b, a%b

qutient, remainer = get_quotient_remainer(x,y)
print('몫: {0}, 나머지: {1}'.format(qutient, remainer))
```

```python
# 심사문제: 사칙연산 함수 만들기
# 표준 입력으로 숫자 2개가 입력됨. 다음 소스 코드를 완성하여 두 숫자의 덧셈, 뺄셈, 곱셈, 나눗셈의
# 결과가 출력되게 만드세요. 이때 나눗셈의 결과는 실수라야 합니다.

x, y = map(int,input().split())

def calc(a,b):
    return a+b, a-b, a*b, a/b

a, s, m ,d = calc(x,y)
print('덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}'.format(a, s, m, d))
```

```python
# 닷셈 함수 만들기

def add(a, b):
    print(a+b)

add(10,50) # 함수를 호출할때 매개변수를 입력을 해야한다.
```

```python
# 함수의 결과를 반환하기
# def 함수이름(매개변수):
#    return(반환값)

def add(a,b):
    return(a+b)
```

```python
add(10,20) # 위에랑 똑같은데 머가 다르냐? 이제 알려줌
```

```python
x = add(10,40)
print(x)
```

```python
# 함수에서 값을 여러개 반환하기
def add_sub(a,b):
    return a+b, a-b

result = add_sub(10,20)
print(result)
```

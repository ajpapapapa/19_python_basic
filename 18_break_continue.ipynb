```python
# while에서 break로 반복문 끝내기

i=0
while True:
    print(i)
    i += 1
    if i == 100:
        break
```

```python
# for에서 break로 반복문 끝내기

for i in range(100):
    print(i)
    if i == 10:
        break
```

```python
# continue로 코드 실행 건너뛰기
for i in range(100):
    if i % 2 == 0: # i를 2로 나머지가 0면 짝수
        continue # print 가기 전에 다시 조건문으로 돌아 가는 것, 조건에 맞으면 print 실행으로 간다.
    print(i)
```

```python
i=0
while i < 100:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
count = int(input())

i = 0
while True:
    print(i)
    i += 1
    if i == count:
        break
```

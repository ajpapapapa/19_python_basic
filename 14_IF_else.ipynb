```python
# 14.7 심사문제 : 합격 여부 판단하기
# 표준 입력으로 국어, 영어, 수학, 과학 점수가 입력됩니다.
# 여기서 네 과목의 평균 점수가 80점 이상일 때 합격이라고 정했습니다.
# 평균 점수에 따라 '합격', '불합격'을 출력하는 프로그램을 만드세요
# 단, 점수는 0점부터 100점까지만 입력받을 수 있으며 범위를 벗어났다면 '잘못된 점수'를
# 출력하고 합격, 불합격 여부는 출력하지 않아야 합니다.

k=int(input())
e=int(input())
m=int(input())
s=int(input())
# kor, eng, math, sci = map(int, input().split())

# 1. 점수가 0~100 범위를 벗어났는지 먼저 검사 (하나라도 벗어나면 잘못된 점수)
if (k < 0 or k > 100) or (e < 0 or e > 100) or (m < 0 or m > 100) or (s < 0 or s > 100):
    print('잘못된 점수')

# 2. 범위가 정상일 때 평균 계산 후 합격/불합격 판정
else:
    average = (k + e + m + s) / 4
    
    if average >= 80:
        print('합격')
    else:
        print('불합격')
```

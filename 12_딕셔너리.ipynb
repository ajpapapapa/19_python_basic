```python
# 딕셔너리의 키에 접근하고 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health'] # 키값을 적으면 벨류값을 가지고 온다고 생각하면 좋다.
print(lux['health'])
print(lux['armor'])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health'] = 2037

lux['mana_regen'] = 3.28

lux
```

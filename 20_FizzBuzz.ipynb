```python
# 실수 많이 하는 과정
for i in range(1,101):
    if i % 3 == 0:
        print('Buzz') # i가 30이면 buzz를 출력하고 그냥 넘어가버림.
    elif i % 5 == 0:
        print('Fizz')
    elif i % 3 == 0 and i % 5 == 0: # 3과 5의 공배수는 검사하지 못함.
        print('FizzBuzz')
    else:
        print(i)
```

```python
# 20.7 연습문제: 2와 11의 배수, 공배수 처리하기
# 1부터 100까지 숫자를 출력하면서 2 배수는 Fizz, 11 배수는 Buzz, 공배수는 FizzBuzz

for i in range(1,101):
    if i%2==0 and i%11==0:
        print('FizzBuzz')
    elif i%2==0:
        print('Fizz')
    elif i%11==0:
        print('Buzz')
    else:
        print(i)
```

```python
# 20.8 심사문제: 5와 7의 배수, 공배수 처리하기
# 표준입력으로 정수 두 개 입력(첫 1~1000, 두 10~1000이며 첫번쨰가 두번째보다 항상작다)
# 5 배수는 Fizz, 7 배수는 Buzz, 공배수는 FizzBuzz

first, second = map(int, input().split())

for i in range(first,second+1):
    if i%5==0 and i%7==0:
        print('FizzBuzz')
    elif i%5==0:
        print('Fizz')
    elif i%7==0:
        print('Buzz')
    else:
        print(i)
```

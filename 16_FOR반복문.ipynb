```python
# for 와 range 사용하기
# for 변수 in range():
#     반복할 코드

for i in range(100): # i는 임의 변수 100은 반복횟수
    print("Hello, world!")
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(10): # i는 임의 변수 100은 반복횟수
    print("Hello, world!",i) 
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기
for a in range(5,12):
    print("hello",a)
```

```python
for i in reversed(range(10)):
    print(i)
```

```python
# 시퀀스 객체로 반복하기
a = [10,20,30,40,50]
for num in a:
    print(num)
```

```python
fruits = ('apple', 'orange', 'grape')
for fruit in fruits:
    print(fruit)
```

```python
# 16.5 연습문제 : 리시트의 요소에 10을 곱해서 출력하기
x = [49,-17,25,102,8,62,21]

for i in x:
    print(i * 10, end = ' ')
```

```python
# 16.6 심사문제: 구구단 출력하기

#표준 입력으로 정수가 입력됩니다. 입력된 정수의 구구단을 출력하는 프로그램을 만드세요
#출력 형식은 숫자 * 숫자 = 숫자처럼 만들고 숫자와 *, = 사이는 공백을 한 칸 띄웁니다.

x = int(input())

for i in range(1,10):
    print(x, '*', i, '=', x*i)
```
